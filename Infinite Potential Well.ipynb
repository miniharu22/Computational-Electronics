```python
# Import numpy module
import numpy as np

# Generate Discrete Laplacian Matrix  
N = 48

A = np.zeros((N, N))
np.fill_diagonal(A, -2)

for ii in range(0, N-1):
    A[ii, ii+1] = 1
    A[ii+1, ii] = 1

A
```

```python
# Extract Eigenvalue & Eigendata in Scaled Potentail Well
D, V = np.linalg.eig(A)

D, V
```

```python
# Sort in ascending order based on eigenvalues
np.argsort(-D)
```

```python
# Check the components of the ground state
-D[30]
```

```python
# Import matplotlib module
import matplotlib.pyplot as plt

# Visualize Wave function based on the the ground state
plt.plot(V[:,30])

plt.show
```
